```python
df = pd.read_csv('survey_results_public.csv', low_memory=False)
df1 = pd.read_csv('survey_results_schema.csv')
print(f"Завантажено рядків: {df.shape[0]}")
df.head()
```

```python
# Завдання 1. Підрахунок загальної кількості респондентів перший спосіб

total_respondents = df.shape[0]
total_respondents
```

```python
# Завдання 1. Підрахунок загальної кількості респондентів другий спосіб

total_respondents = df['ResponseId'].nunique()
total_respondents
```

```python
# Завдання 2. Аналіз повноти відповідей респондентів

questions_set = set(df1['qname'].unique())
common_columns = list(questions_set.intersection(df.columns))
clean_df = df.dropna(subset=common_columns)
total_complete = clean_df.shape[0]

print(f"Кількість респондентів, які відповіли на всі питання: {total_complete}")
```

```python
# Завдання 3. Статистичний аналіз досвіду респондентів

work_exp_mean = df['WorkExp'].mean()
work_exp_median = df['WorkExp'].median()
work_exp_mode = df['WorkExp'].mode()[0]

print(f"Міри центральної тенденції для WorkExp:")
print(f"Середнє значення: {work_exp_mean:.1f}")
print(f"Медіана:          {work_exp_median}")
print(f"Мода:             {work_exp_mode}")
```

```python
# Завдання 4. Аналіз віддаленої роботи
df['RemoteWork'].unique()
```

```python
remote_workers = df[df['RemoteWork'] == 'Remote']
remote_count = remote_workers.shape[0]
remote_count
```

```python
# Завдання 5. Визначення популярності Python

df['worked_with_python'] = df['LanguageHaveWorkedWith'].str.contains('Python', na=False)
python_users_count = df['worked_with_python'].sum()
total_respondents = len(df)
python_percentage = (python_users_count / total_respondents) * 100

print(f"Відсоток респондентів: {python_percentage:.1f}%")
```

```python
# Завдання 6. Аналіз шляхів навчання програмуванню

df['Learned_with_online_courses'] = df['LearnCode'].str.contains('Online Courses', na=False)
online_learners_count = df['Learned_with_online_courses'].sum()

print(f"Кількість респондентів, що навчалися через онлайн-курси: {online_learners_count}")
```

```python
#Завдання 7. Географічний аналіз компенсації Python-розробників

python_devs = df[df['worked_with_python'] == True].dropna(subset=['ConvertedCompYearly'])
py_progers_by_country = python_devs.groupby('Country')['ConvertedCompYearly'].agg(['mean', 'median'])
py_progers_by_country
```

```python
# Завдання 8. Аналіз освіти найбільш оплачуваних спеціалістів

top_paid_df = df[['ResponseId', 'EdLevel', 'ConvertedCompYearly']]
top_paid_df = top_paid_df.sort_values(by='ConvertedCompYearly', ascending=False)
top_paid_df = top_paid_df.reset_index(drop=True)
top_5_paid = top_paid_df.head(5)
top_5_paid
```

```python
# Завдання 9. Аналіз популярності Python по віковим категоріям

age_python_stats = df.groupby('Age')['worked_with_python'].mean() * 100
age_python_stats = age_python_stats.reset_index()
age_python_stats.columns = ['Вікова категорія', 'Відсоток Python-розробників']
age_python_stats['Відсоток Python-розробників'] = age_python_stats['Відсоток Python-розробників'].round(1)
age_python_stats
```

```python
# Завдання 10. Аналіз індустрій серед високооплачуваних віддалених працівників


income_threshold = df['ConvertedCompYearly'].quantile(0.75)
high_paid_remote = df[
    (df['ConvertedCompYearly'] > income_threshold) &
    (df['RemoteWork'] == 'Remote')
]
high_paid_remote_industries = high_paid_remote['Industry'].dropna()
industry_counts = high_paid_remote_industries.value_counts()

print(f"75-й перцентиль компенсації: {income_threshold:.0f}")
print("-" * 30)
print("Найпопулярніші індустрії серед високооплачуваних віддалених працівників:")
industry_counts.head(10)
```
